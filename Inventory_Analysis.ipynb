PART 3. Clean the Data

```python
#Handle missing values: decide whether to fill or drop them.
df["Product_Name"] = df["Product_Name"].fillna("Unknown")
df
```

```python
#Remove duplicate rows.
df.drop_duplicates()
```

```python
#Standardize text: make all Product_Name, Category, and Supplier lowercase (or title case).
df["Product_Name"] = df["Product_Name"].str.title()
df["Category"] = df["Category"].str.title()
df["Supplier"] = df["Supplier"].str.title()
df
```

```python
#Fix data types: convert columns to proper numeric or datetime types.
# Convert numeric columns
df["Quantity"] = pd.to_numeric(df["Quantity"], errors='coerce')
df["Unit_Price"] = pd.to_numeric(df["Unit_Price"], errors='coerce')
df 
```

```python
# Convert dates
df["Last_Update"] = pd.to_datetime(df["Last_Update"], errors='coerce')
df
```

```python
#Handle incorrect values:
#Replace or drop rows with negative or unrealistic prices/quantities.
df.loc[df["Quantity"] < 0, "Quantity"] = np.nan
df
```

```python
# Remove negative prices
df.loc[df["Unit_Price"] < 0, "Unit_Price"] = np.nan
df
```

```python
# Fill remaining NaN values
df["Quantity"] = df["Quantity"].fillna(df["Quantity"].median())
df["Unit_Price"] = df["Unit_Price"].fillna(df["Unit_Price"].median())
df
```

```python
#Standardize date formats in Last_Updated.
df["Last_Update"] = pd.to_datetime(df["Last_Update"], errors='coerce')
df
```
